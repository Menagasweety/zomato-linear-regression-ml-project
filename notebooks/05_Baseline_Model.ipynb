# 05 — Baseline Model
## Goal
Create a simple reference point before Linear Regression.

A baseline answers:
> "How good is our ML model compared with a very simple prediction?"

We use the training-set mean as the prediction for every test observation.

### Metrics
- **MAE:** average absolute error in minutes.
- **RMSE:** penalizes larger errors more strongly.
- **R²:** proportion of target variance explained relative to the mean baseline.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

df = pd.read_csv("../outputs/cleaned/zomato_model_dataset.csv")

target = "Time_taken (min)"
X = df.drop(columns=[target])
y = df[target]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

baseline_prediction = np.repeat(y_train.mean(), len(y_test))

baseline_metrics = {
    "Model": "Mean Baseline",
    "MAE": mean_absolute_error(y_test, baseline_prediction),
    "RMSE": np.sqrt(mean_squared_error(y_test, baseline_prediction)),
    "R2": r2_score(y_test, baseline_prediction)
}

pd.DataFrame([baseline_metrics]).round(3)
```

### Interpretation

If Linear Regression cannot beat this baseline, there is a serious modeling problem.

The baseline is deliberately simple. It gives us a minimum standard that the ML pipeline should improve upon.
